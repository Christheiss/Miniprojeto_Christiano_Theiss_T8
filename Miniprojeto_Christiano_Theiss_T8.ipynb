```python
import pandas as pd

# Ler o arquivo usando ponto e vírgula como separador
df = pd.read_csv('/content/Base Varejo.csv', sep=';')

# Mostrar as primeiras linhas
display(df.head())

# Diagnóstico da estrutura
print("Quantidade de registros:", df.shape[0])
print("Quantidade de colunas:", df.shape[1])

print("\nNomes das colunas:")
print(df.columns.tolist())

print("\nTipos de dados:")
print(df.dtypes)
```

```python
# 1. Verificar valores ausentes por coluna
print("VALORES AUSENTES POR COLUNA:")
print(df.isnull().sum())

# 2. Verificar textos vazios ou com apenas espaços
print("\nTEXTOS VAZIOS OU COM ESPAÇOS:")
for coluna in df.select_dtypes(include='object').columns:
    vazios = df[coluna].str.strip().eq('').sum()
    print(f"{coluna}: {vazios}")

# 3. Verificar linhas totalmente duplicadas
print("\nLINHAS DUPLICADAS:")
print("Total:", df.duplicated().sum())

# 4. Verificar as colunas extras
print("\nCOLUNAS EXTRAS:")
extras = [col for col in df.columns if col.startswith('Unnamed:')]
print(df[extras].isnull().sum())

# 5. Verificar os valores únicos das colunas textuais
print("\nVALORES ÚNICOS DAS COLUNAS TEXTUAIS:")
for coluna in df.select_dtypes(include='object').columns:
    print(f"\n{coluna}:")
    print(df[coluna].value_counts(dropna=False).head(10))
```

```python
# Trabalhar com uma cópia para investigar a base
df_analise = df.copy()

# Remover somente as colunas totalmente vazias
colunas_vazias = df_analise.columns[
    df_analise.isnull().all()
].tolist()

df_analise = df_analise.drop(columns=colunas_vazias)

print("Colunas vazias identificadas:", colunas_vazias)
print("Colunas restantes:", df_analise.shape[1])

# Verificar exemplos de linhas duplicadas
duplicadas = df_analise[
    df_analise.duplicated(keep=False)
]

print("\nTotal de linhas envolvidas em duplicações:",
      len(duplicadas))

display(duplicadas.head(10))

# Investigar a categoria #N/D
print("\nRegistros com categoria #N/D:")
display(
    df_analise[df_analise["PR_CAT"] == "#N/D"].head(10)
)

# Conferir os valores de CL_FHL
print("\nValores de CL_FHL:")
print(df_analise["CL_FHL"].value_counts(dropna=False).sort_index())
```

```python
# Verificar se os registros repetidos são idênticos
# considerando as colunas disponíveis

df_analise = df.drop(
    columns=[
        col for col in df.columns
        if col.startswith('Unnamed:')
    ]
).copy()

print("Total de registros:", len(df_analise))
print("Duplicatas exatas:", df_analise.duplicated().sum())

# Examinar um grupo de registros repetidos
exemplo = df_analise[
    df_analise.duplicated(keep=False)
]

display(exemplo.head(20))

# Verificar se há repetições da mesma combinação
# de data, compra, cliente e produto
chaves = ['DATA', 'CO_ID', 'CL_ID', 'PR_ID']

print("\nDuplicatas pela combinação de data, compra, cliente e produto:")
print(df_analise.duplicated(subset=chaves).sum())

# Verificar os registros do produto 107
print("\nRegistros do produto 107:")
display(
    df_analise[df_analise['PR_ID'] == 107]
    [['PR_ID', 'PR_CAT', 'PR_NOME']]
    .drop_duplicates()
)
```

```python
def tratar_dados(dados):

    """
    Remove colunas totalmente vazias e padroniza textos.
    Mantém duplicatas e categorias desconhecidas
    até haver evidência para removê-las.
    """
    dados = dados.copy()

    # Remover somente colunas completamente vazias
    dados = dados.dropna(axis=1, how='all')

    # Remover espaços extras e padronizar textos
    colunas_texto = dados.select_dtypes(
        include='object'
    ).columns

    for coluna in colunas_texto:
        dados[coluna] = dados[coluna].str.strip()

    # Padronizar nomes de produtos e categorias em maiúsculas
    for coluna in ['PR_CAT', 'PR_NOME', 'CL_GENERO', 'CL_SEG']:
        if coluna in dados.columns:
            dados[coluna] = dados[coluna].str.upper()

    return dados


# Aplicar a função
df_limpo = tratar_dados(df)

print("Registros antes:", len(df))
print("Registros depois:", len(df_limpo))
print("Colunas antes:", df.shape[1])
print("Colunas depois:", df_limpo.shape[1])
print("Duplicatas mantidas:", df_limpo.duplicated().sum())
```

```python
# Estatísticas básicas da coluna CL_FHL
coluna = 'CL_FHL'

print("Quantidade de valores válidos:", df_limpo[coluna].count())
print("Média:", df_limpo[coluna].mean())
print("Mínimo:", df_limpo[coluna].min())
print("Máximo:", df_limpo[coluna].max())

print("\nResumo estatístico completo:")
display(df_limpo[coluna].describe())
```

```python
# FILTRO: selecionar registros da categoria ALIMENTOS
filtro_alimentos = df_limpo[
    df_limpo['PR_CAT'] == 'ALIMENTOS'
]

print("Registros da categoria ALIMENTOS:",
      len(filtro_alimentos))

# AGRUPAMENTO 1: quantidade de registros por categoria
print("\nQuantidade de registros por categoria:")
resumo_categoria = df_limpo.groupby(
    'PR_CAT'
).size().sort_values(ascending=False)

display(resumo_categoria)

# AGRUPAMENTO 2: quantidade de registros por gênero
print("\nQuantidade de registros por gênero:")
resumo_genero = df_limpo.groupby(
    'CL_GENERO'
).size().sort_values(ascending=False)

display(resumo_genero)
```

```python
# Quantidade de registros antes e depois da limpeza
print("Registros antes:", len(df))
print("Registros depois:", len(df_limpo))

# Conferir problemas restantes
print("\nValores ausentes por coluna:")
print(df_limpo.isnull().sum())

print("\nDuplicatas mantidas:")
print(df_limpo.duplicated().sum())

# Exportar a base tratada sem o índice
df_limpo.to_csv(
    '/content/varejo_limpo.csv',
    index=False,
    sep=';',
    encoding='utf-8-sig'
)

print("\nArquivo varejo_limpo.csv exportado com sucesso!")
```

## 3. Estrutura da base

A base original contém 830.000 registros e 14 colunas. Após a remoção de quatro colunas completamente vazias, restaram 830.000 registros e 10 colunas.

As colunas utilizadas são:

- `DATA`: data registrada.
- `CO_ID`: identificador associado à compra ou operação, conforme a estrutura da base.
- `CL_ID`: identificador do cliente.
- `CL_GENERO`: gênero registrado do cliente.
- `CL_EC`: variável cadastral do cliente; o significado exato deve ser confirmado na documentação da base.
- `CL_FHL`: variável numérica do cliente, com valores de 0 a 4; o significado exato deve ser confirmado.
- `CL_SEG`: segmento do cliente.
- `PR_ID`: identificador do produto.
- `PR_CAT`: categoria do produto.
- `PR_NOME`: nome do produto.

A unidade de análise adotada é o registro da base. Portanto, as contagens apresentadas não devem ser interpretadas automaticamente como clientes únicos, compras únicas ou faturamento.

## 4. Diagnóstico e decisões de limpeza

O diagnóstico inicial identificou:

- 830.000 registros.
- Nenhum valor ausente nas dez colunas principais.
- Quatro colunas extras (`Unnamed: 10` a `Unnamed: 13`) completamente vazias.
- Nenhum texto vazio ou composto apenas por espaços nas colunas textuais verificadas.
- 96.553 linhas duplicadas exatas.
- 3.650 registros com `#N/D` nas categorias de produto.

As quatro colunas totalmente vazias foram removidas. Os textos das colunas categóricas foram padronizados com remoção de espaços nas extremidades e conversão para maiúsculas.

As linhas duplicadas foram mantidas, pois a base não possui uma coluna de quantidade que permita distinguir com segurança uma repetição indevida de itens repetidos em uma compra. O valor `#N/D` também foi mantido, pois não há documentação suficiente para determinar se deve ser convertido em valor ausente ou excluído.

## 5. Processo de tratamento

A leitura da base foi realizada com `pd.read_csv()`, utilizando `sep=';'` para separar corretamente as colunas. A transformação foi implementada na função `tratar_dados()`, que remove colunas totalmente vazias e padroniza textos. A função `gerar_resumo()` foi utilizada para produzir resumos categóricos.

A base tratada foi exportada para `varejo_limpo.csv`, sem índice, usando `index=False`. A limpeza removeu quatro colunas sem informação, mas preservou os 830.000 registros, evitando exclusões sem justificativa.

## 6. Resultados e conclusões

- A base possui 830.000 registros e 10 colunas úteis após a remoção das quatro colunas vazias.
- A categoria `ALIMENTOS` apresenta 434.767 registros, aproximadamente 52,4% do total.
- A categoria `HIGIENE` apresenta 155.574 registros, seguida por `LIMPEZA`, com 145.754.
- Os registros associados ao gênero `F` somam 432.576, enquanto os associados a `M` somam 397.424.
- A variável `CL_FHL` apresenta 830.000 valores válidos, média de aproximadamente 1,1465, mínimo 0 e máximo 4.
- Foram encontradas 96.553 linhas duplicadas exatas e 3.650 registros com `#N/D`. Esses casos foram mantidos para evitar exclusões sem evidência suficiente.
